```python
from huggingface_hub import hf_hub_download
from pathlib import Path

# Read token from file
with open("HF_TOKEN.txt", "r") as f:
    token = f.read().strip()

# Custom download directory
custom_dir = Path("../codes/diffusion_sd1/stable-diffusion/models/ldm/stable-diffusion-v1/")
custom_dir.mkdir(parents=True, exist_ok=True)

# Download file
path = hf_hub_download(
    repo_id="CompVis/stable-diffusion-v-1-4-original",
    filename="sd-v1-4.ckpt",
    token=token,                 # new versions use `token` instead of use_auth_token
    local_dir=custom_dir,        # download location
    local_dir_use_symlinks=False # ensures real file copy
)

print("Downloaded to:", path)
```
